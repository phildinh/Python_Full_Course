```python
import pandas as pd
import numpy as np
import json
```

```python
df_main = pd.read_excel("MentorMe-CV-File-20260224.xlsx",sheet_name="Main_table")
df_user = pd.read_csv("dim_user.csv")
```

```python
# Open dim customer to see the columns again
df_user.head()
```

```python
# How many nulls in each potential join column in df_main?
print(df_main[['candidate_name', 'email']].isnull().sum())

# How many nulls in dim_customer join columns?
print(df_user[['candidate_name', 'email']].isnull().sum())
```

We need to join both table to get normalize customer_id continuing to generate different table for analysis

```python
# Normalize candidate name and email from main table
df_main['candidate_name'] = df_main['candidate_name'].str.lower().str.strip()
df_main['email'] = df_main['email'].str.lower().str.strip()
```

```python
# We try to join 2 tables first then continue the work
df_merge = df_main.merge(df_user[['customer_id','candidate_name','email']],
                         on=['candidate_name','email'],
                         how='left')
df_merge.head()
```

```python
# Check null value for id after merging
df_merge['customer_id'].isnull().sum()
```

```python
# Show rows where email is null — did they get a customer_id?
df_merge[df_merge['email'].isnull()][['candidate_name', 'email', 'customer_id']]
```

```python
# now we check our dataset with customer_id
df_merge.head(3).T
```

```python
# We extract to skills table then work from here
df_skill = df_merge[['customer_id','analysis_id','technical_skills','soft_skills']].copy()
df_skill
```

```python
# Before we dive in we will check the value from technical and sofk skills
print("We check first row for tech skills and data type")
print(df_skill['technical_skills'].iloc[0])
print(type(df_skill['technical_skills'].iloc[0]))
print("\n")

print("We check first row for soft skills and data type")
print(df_skill['soft_skills'].iloc[0])
print(type(df_skill['soft_skills'].iloc[0]))
print("\n")
```

```python
# We can understand right now data type is str and data structure for skill columns are list of dictionaries
# We will use small row to work with, if it work will apply to the rest of them
raw = df_skill['technical_skills'].iloc[0]
parsed = json.loads(raw)

print(type(parsed))
print(parsed[1])
```

```python
df_test = pd.DataFrame(parsed)
df_test
```

```python
results = []

for index,row in df_skill.iterrows():
    # Step 1: parse the json string
    parsed = json.loads(row['technical_skills'])

    # step 2: convert to dataframe
    temp_df = pd.DataFrame(parsed)

    #Step 3: add customer_id and analysis_id to everyrow
    temp_df['customer_id'] = row['customer_id']
    temp_df['analysis_id'] = row['analysis_id']
    temp_df['skill_type'] = 'technical_skills'
    results.append(temp_df)

    # step 4: Append the result
    df_tech = pd.concat(results).reset_index(drop=True)
    print(df_tech)
    print(type(df_tech))
```

```python
results_2 = []

for index,row in df_skill.iterrows():
    # Step 1: parse the json string
    parsed = json.loads(row['soft_skills'])

    # step 2: convert to dataframe
    temp_df = pd.DataFrame(parsed)

    #Step 3: add customer_id and analysis_id to everyrow
    temp_df['customer_id'] = row['customer_id']
    temp_df['analysis_id'] = row['analysis_id']
    temp_df['skill_type'] = 'soft_skills'
    results_2.append(temp_df)

    # step 4: Append the result
    df_soft = pd.concat(results_2).reset_index(drop=True)
    print(df_soft)
    print(type(df_soft))
```

```python
df_skills_final = pd.concat([df_soft,df_tech]).reset_index(drop=True)
df_skills_final
```

```python
df_skills_final.to_csv("dim_skills.csv", index = False)
```
