```python
import pandas as pd
from sqlalchemy import create_engine,text, BigInteger
```

```python
# read CSV
df_customer = pd.read_csv("../data/dim_user.csv")
df_fact = pd.read_csv("../data/fact_cv_analysis.csv")
df_skills = pd.read_csv("../data/dim_skills.csv")

# Fix fact_cv_analysis columns
df_fact = df_fact.rename(columns={
    'experience_level.level' : 'experience_level',
    'experience_level.years' : 'experience_years'
})

# Fix dim_skills columns
df_skills = df_skills.rename(columns={
    'name'  : 'skill_name',
    'level' : 'skill_level',
    'score' : 'skill_score'
})
```

```python
# Connection
engine = create_engine(
    "mssql+pyodbc://PHIL\\SQLEXPRESS/mentorme"
    "?driver=ODBC+Driver+17+for+SQL+Server"
    "&trusted_connection=yes"
)

try:
# Full Load
    with engine.connect() as conn:
        # children first
        conn.execute(text("DELETE FROM dim_skills"))
        conn.execute(text("DELETE FROM fact_cv_analysis"))
        conn.execute(text("DELETE FROM dim_customer"))
        conn.commit()
        print("✅ Tables cleared successfully!")

    # parents first
    df_customer.to_sql(
        "dim_customer",
        con       = engine,
        if_exists = "append",
        index     = False
    )

    df_fact.to_sql(
        "fact_cv_analysis",
        con       = engine,
        if_exists = "append",
        index     = False,
        dtype     = {"analysis_id": BigInteger()}
    )

    df_skills.to_sql(
        "dim_skills",
        con       = engine,
        if_exists = "append",
        index     = False,
        dtype     = {"analysis_id": BigInteger()}
    )
    
    print(f"✅ dim_customer    → {len(df_customer)} rows")
    print(f"✅ fact_cv_analysis → {len(df_fact)} rows")
    print(f"✅ dim_skills       → {len(df_skills)} rows")
    print("\n🎉 Full load completed successfully!")

except Exception as e:
    print(f"Full Load failed: {e}")
```

```python
df_customer.to_sql("dim_customer", con=engine, if_exists="append", index=False)
print("dim_customer ✅")
```

```python
df_fact.to_sql("fact_cv_analysis", con=engine, if_exists="append", index=False, dtype={"analysis_id": BigInteger()})
print("fact_cv_analysis ✅")
```

```python
df_skills.to_sql("dim_skills", con=engine, if_exists="append", index=False, dtype={"analysis_id": BigInteger()})
print("dim_skills ✅")
```
