```python
import requests
import json
import os
from dotenv import load_dotenv
```

```python
response = requests.get(
    url="https://jsonplaceholder.typicode.com/todos",
    params={
        "_limit": 10,
        "_start": 0
    }
)

response.raise_for_status()
data = response.json()

print(f"Records returned: {len(data)}")
print(json.dumps(data[0], indent=2))  # print just the first record
```

```python
all_todos = []
start = 0
limit = 10

while True:
    response = requests.get(
        url = "https://jsonplaceholder.typicode.com/todos",
        params = {
            "_limit": limit,
            "_start": start
        }
    )

    response.raise_for_status()

    batch = response.json()

    if not batch:
        break
    all_todos.extend(batch)
    print(f"Fetched records {start + 1} to {start + len(batch)}")
    start += limit

print(f"\n Total todos fetched: {len(all_todos)}")
```

```python
response = requests.get(
    url="https://rickandmortyapi.com/api/character",
    params={"page": 1}
)

response.raise_for_status()
print(json.dumps(response.json(), indent = 2))
```

```python
all_characters = []
page = 1

while True:
    try:
        response = requests.get(
            url = "https://rickandmortyapi.com/api/character",
            params = {"page": page}
        )
        response.raise_for_status()
        data = response.json()

        all_characters.extend(data["results"])
        print(f"Fetched page {page} — {len(data['results'])} characters")

        if not data["info"]["next"]:
            break

        page += 1

    except requests.exceptions.HTTPError as e:
        print(f"HTTP Error: {e.response.status_code} - {e.response.text}")
        break

    except requests.exceptions.ConnectionError:
        print("Connection error — could not reach the API. Check your network.")
        break

    except requests.exceptions.Timeout:
        print("Timeout — the API took too long to respond.")
        break

    except requests.exceptions.RequestException as e:
        print(f"Unexpected error: {e}")
        break

print(f"Total characters fetched: {len(all_characters)}")
```
