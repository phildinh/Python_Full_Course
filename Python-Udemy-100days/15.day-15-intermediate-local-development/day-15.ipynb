```python
import random

# ===== Functions =====

def coffee_option():
    while True:
        choice = input("What would you like? (espresso/latte/cappuccino): ").lower().strip()
        if choice in ["espresso", "latte", "cappuccino"]:
            return choice
        print("Please choose again.")

def turn_off():
    while True:
        choice_turn_off = input("Turn off the machine for maintenance? (y/n): ").lower().strip()
        if choice_turn_off == 'y':
            return False
        if choice_turn_off == 'n':
            return True
        print("Please type 'y' or 'n'.")

def print_report(water, milk, coffee, money):
    print(f"Current Water: {water}ml")
    print(f"Current Milk: {milk}ml")
    print(f"Current Coffee: {coffee}g")
    print(f"Current Money: ${money}")

def input_money():
    while True:
        try:
            money_in = int(input("Insert note (5/10/20 AUD only): "))
        except ValueError:
            print("Please enter a number (5, 10, or 20).")
            continue

        if money_in in (5, 10, 20):
            return money_in
        print("Only 5, 10, 20 AUD notes are accepted.")

def continue_order():
    while True:
        choice_continue = input("Would you like to order more? (y/n): ").lower().strip()
        if choice_continue == 'y':
            return True
        if choice_continue == 'n':
            return False
        print("Please type 'y' or 'n'.")

def has_resources(choice, water, milk, coffee, recipes):
    need = recipes[choice]["ingredients"]
    if water < need.get("water", 0):
        return False
    if milk < need.get("milk", 0):
        return False
    if coffee < need.get("coffee", 0):
        return False
    return True

def process_payment(price, machine_money):
    customer_money = input_money()
    change = customer_money - price

    if change < 0:
        print("Not enough money. Money refunded.")
        return None, machine_money

    # Machine must have enough cash to give change
    if machine_money < change:
        print("Sorry, we don't have enough change. Money refunded.")
        return None, machine_money

    # Machine keeps the price, returns change
    machine_money += price
    machine_money -= change

    return change, machine_money


# ===== Data =====

recipes = {
    "espresso": {
        "price": 2,
        "ingredients": {"water": 50, "coffee": 18}
    },
    "latte": {
        "price": 3,
        "ingredients": {"water": 200, "milk": 150, "coffee": 24}
    },
    "cappuccino": {
        "price": 4,
        "ingredients": {"water": 250, "milk": 100, "coffee": 24}
    }
}

water = 500
milk = 500
coffee = 200
money = 100
coffee_working = True


# ===== Main Loop =====

while coffee_working:
    print("\nWelcome to the coffee machine")

    customer_choice = coffee_option()

    # 1) Check resources first (do NOT subtract yet)
    if not has_resources(customer_choice, water, milk, coffee, recipes):
        print("Sorry, we are out of materials.")
        print_report(water, milk, coffee, money)
        coffee_working = turn_off()
        continue

    price = recipes[customer_choice]["price"]

    # 2) Take payment
    change, money = process_payment(price, money)

    if change is None:
        # Payment failed (not enough money or not enough change)
        coffee_working = turn_off()
        continue

    # 3) Now subtract ingredients (only after successful payment)
    need = recipes[customer_choice]["ingredients"]
    water -= need.get("water", 0)
    milk -= need.get("milk", 0)
    coffee -= need.get("coffee", 0)

    # 4) Deliver coffee
    print(f"Here is your {customer_choice} ☕")
    print(f"This is your change: AUD {change}")
    print("Your coffee is in progress, please wait up to 5 minutes")
    print_report(water, milk, coffee, money)

    # 5) Continue?
    coffee_working = continue_order()

print("Machine stopped. Goodbye!")
```
