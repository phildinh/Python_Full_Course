```python
from turtle import Turtle

# Turtle is class
timmy = Turtle()
print(timmy)
timmy.shape("turtle")
timmy.color("coral")
timmy.forward(100)
```

```python
from turtle import Turtle, Screen

# Turtle is class
timmy = Turtle()
print(timmy)
timmy.shape("turtle")
timmy.color("coral")
timmy.forward(100)

# Screen() is a class, we have my_screen is an object then attribute is canvheight
my_screen = Screen()
print(my_screen.canvheight)
my_screen.exitonclick()
```
