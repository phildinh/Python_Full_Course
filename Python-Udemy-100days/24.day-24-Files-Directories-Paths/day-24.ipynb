```python
file = open("my_file.txt")
contents = file.read()
print(contents)
file.close()
```

```python
with open("my_file.txt") as file:
    contents = file.read()
    print(contents)
```

```python
# Remove and write all new file
with open("my_file.txt", mode = "w") as file:
    file.write("New text.")
```

```python
# Append task to file
with open("my_file.txt", mode = "a") as file:
    file.write("New text.")
```
