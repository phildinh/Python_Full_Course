```python
word_list = ["aardvark", "baboon","camel"]

import random

# TODO-1 -Randomly choose a word from the word_list and assign it to a variable called chosen_word. Then print it.

chosen_word = random.choice(word_list)
print(chosen_word)

# TODO-1.1 Create a "placeholder" with the same number of blanks as the chosen_word

placeholder = ""
for r in range(1, len(chosen_word) + 1):
    placeholder += "_"

# TODO-2 Ask the user to guess a letter and assign their answer to a variable called guess. Make guess lowercase.

game_over = False
correct_letters = []

while not game_over:
    guess = input("Please pick one letter from your keyboard for our game").lower()
    print(f"Guess a letter: {guess}")
# TODO-3 Check if the letter the user guessed (guess) is one of the letters in the chosen_Word. Print "Right" if it is, "Wroing" if it is not.
    display = ""
    for letter in chosen_word:
        if letter == guess:
            display += letter
            correct_letters.append(letter)
        elif letter in correct_letters:
            display += letter
        else:
            display += "_"
    print(display)
    if "_" not in display:
        game_over = True
        print("You Win!")
```
