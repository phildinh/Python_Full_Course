```python
# Pizza challenge:
print("Welcome to Python Pizza Deliveries!")
size = input("What size pizza do you want? S, M or L: ")
pepperoni = input("Do you want pepperoni on your pizza? Y or N: ")
extra_cheese = input("Do you want extra cheese? Y or N: ")

bill = 0

if size == "S":
    bill = 15
    print("Small pizza is $15 ")
    if pepperoni == "Y":
        bill += 2
        if extra_cheese == "Y":
            bill += 1
            print(f"Total bill for small pizza with extra pepperoni and cheese is ${bill}")
    elif extra_cheese == "Y":
        bill += 1
        print(f"Total bill for small pizza with extra cheese is ${bill}")

elif size == "M":
    bill = 20
    print("Medium pizza is $20 ")
    if pepperoni == "Y":
        bill += 3
        if extra_cheese == "Y":
            bill += 1
            print(f"Total bill for medium pizza with extra pepperoni and cheese is ${bill}")
    elif extra_cheese == "Y":
        bill += 1
        print(f"Total bill for medium pizza with extra cheese is ${bill}")
else:
    bill = 25
    print("Large pizza is $25 ")
    if pepperoni == "Y":
        bill += 3
        if extra_cheese == "Y":
            bill += 1
            print(f"Total bill for medium pizza with extra pepperoni and cheese is ${bill}")
    elif extra_cheese == "Y":
        bill += 1
        print(f"Total bill for medium pizza with extra cheese is ${bill}")
```
