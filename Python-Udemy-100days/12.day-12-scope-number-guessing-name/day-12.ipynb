```python
# If I create a variable within a function it is local scope
# If I create a variable just outside like normal or in if condition, it is a global scope
```

```python
# Variable in if condition is global
game_level = 3
enemies = ["a","b","c"]

if game_level < 5:
    new_enemy = enemies[0]

print(new_enemy)
```

```python
# Variable in def is local scope
game_level = 3

def game():
    game_level = 2
    print(f"game level in function {game_level}")

game()
print(game_level)
```

```python
# There is no block scope. Inside a if/else/for/whike code block is the same as outside it
def bar():
    a = 100

    if 20 > 5:
        a = 200
        print(a)

bar()
```

```python
# Modifying Global Scope, but very becareful because it can cause many errors

enemies = 1 

def increase_enemies():
    global enemies
    enemies += 1
    print(f"enemies inside function: {enemies}")

increase_enemies()
print(f"enemies outside function {enemies}")
```

```python
# Other way not to use global scope:
enemies = 1 

def increase_enemies(enemies):
    print(f"enemies inside function: {enemies}")
    return enemies + 1

enemies = increase_enemies(enemies)
print(f"enemies outside function {enemies}")
```

```python
# Project
import random

number = random.randint(1,100)

print("Welcome to the Number Guessing Game!")
print("I'm thinking of a number between 1 and 100.")
level = input("Choose a difficulty. Type 'easy' or 'hard'").lower()

easy_life = 10
hard_life = 5

while True:

    if level == "hard":
        print(f"you have {hard_life} attempts remaining to guess the number")
        guess = int(input("Make a guess!"))
        if guess != number:
            hard_life -= 1
            if guess > number:
                print("Too high !")
            else:
                print("Too Low !")
            if hard_life == 0:
                print("You loose all your attempts and Lose!")
                print(f"The correct number is {number}")
                break

        else:
            print("You guess the correct number!")
            break

    elif level == "easy":
        print(f"you have {easy_life} attempts remaining to guess the number")
        guess = int(input("Make a guess"))
        if guess != number:
            easy_life -= 1
            if guess > number:
                print("Too high !")
            else:
                print("Too Low !")           
            if easy_life == 0:
                print("You loose all your attempts and Lose!")
                print(f"The correct number is {number}")
                break

        else:
            print("You guess the correct number!")
            break       

    else:
        print("Please try again and choose a correct level")
        break
```
