```python
student_scores = {
    'Harry': 88,
    'Ron': 78,
    'Hermione': 95,
    'Draco': 75,
    'Neville': 60
}

student_grades = {}

for k,v in student_scores.items():

    if 91 <= v <= 100:
        student_grades[k] = "Outstanding"
    elif 81 <= v <= 90:
        student_grades[k] = "Exceeds Expectations"
    elif 71 <= v <= 80:
        student_grades[k] = "Acceptable"
    else:
        student_grades[k] = "Fail"  

print(student_grades)
```

```python
running = True
house_auction = {}


while running:
    
    name = input("What is your name: ").lower()
    bid = input("what is your bid: ")

    if name in house_auction:
        print("You already placed your bid, updating your bid....")

    house_auction[name] = bid

    answer = input("If you want to continue bid type 'Yes', if not type 'No'").lower()

    print(house_auction)

    if answer == "no":
        winner = max(house_auction, key=house_auction.get)
        highest_bid = house_auction[winner]
        print(f"Winner: {winner}, Bid: {highest_bid}")
        running = False

```

```python
house_auction = {
    "harry": 300,
    "ron": 250,
    "hermione": 500,
    "draco": 400
}

highest_bid = 0
winner = {}
for k,v in house_auction.items():
    if v > highest_bid:
        highest_bid = v
        winner = {k:v}

print(winner)
```
