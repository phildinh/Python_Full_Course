```python
class User:
    
    def __init__(self,user_id,username):
        print("new user being created..")
        self.id = user_id
        self.username = username
        self.followers = 0
        self.following = 0

    def follow(self, user):
        user.followers += 1
        self.following += 1

user_1 = User("001","Phil")
user_2 = User("002","Phil2")

user_1.follow(user_2)
print(user_1.followers)
print(user_1.following)
print(user_2.followers)
print(user_2.following)
```

```python
quiz_data = [
    {
        "question": "Python is a programming language.",
        "answer": "True"
    },
    {
        "question": "Sydney is the capital of Australia.",
        "answer": "False"
    },
    {
        "question": "A dictionary in Python stores data as key-value pairs.",
        "answer": "True"
    },
    {
        "question": "The Earth has two moons.",
        "answer": "False"
    },
    {
        "question": "SQL is used to manage databases.",
        "answer": "True"
    },
    {
        "question": "Lists in Python are immutable.",
        "answer": "False"
    },
    {
        "question": "The Boolean data type has only two values.",
        "answer": "True"
    },
    {
        "question": "HTML is a programming language.",
        "answer": "False"
    }
]
```

```python
class Question:
    def __init__(self,q_question,q_answer):
        self.question = q_question
        self.answer = q_answer
        
class QuizBrain:

    def __init__(self, q_list):
        self.question_number = 0
        self.score = 0
        self.question_list = q_list

    def still_has_question(self):
        return self.question_number < len(self.question_list)
    
    def next_question(self):
        current_question = self.question_list[self.question_number]
        self.question_number += 1
        user_answer = input(
            f"Q.{self.question_number}: {current_question.question} (True/False): "
        )
        self.check_answer(user_answer,current_question.answer)

    def check_answer(self,user_answer,correct_answer):
        if user_answer.lower() == correct_answer.lower():
            self.score +=1
            print("You got it right!")
        else:
            print("That's wrong. ")
        print(f"the correct answer was: {correct_answer}. ")
        print(f"Your current score is: {self.score}/{self.question_number}")
        print("-" * 10)
        print("\n")
```

```python
question_blank = []

for question in quiz_data:
    quiz_question = question["question"]
    quiz_answer = question["answer"]
    new_question = Question(quiz_question,quiz_answer)
    question_blank.append(new_question)

quiz = QuizBrain(question_blank)

while quiz.still_has_question():
    quiz.next_question()

print("You've competed the quiz")
print(f"Your final score was: {quiz.score}/{quiz.question_number}")
```

```python
print("Welcome to the Quiz Game")

score = 0
total_questions = len(quiz_data)

while True:

    for n in range(total_questions):
        current_question_num = n+1
        print(f"This is question number {n+1}: {quiz_data[n]['question']}")
        answer = input("Please choose your answer 'True' or 'False'").strip().lower()
        if answer == quiz_data[n]["answer"].lower():
            print("You got it right!")
            print(f"The correct answer was {quiz_data[n]['answer']}")
            score += 1
            print(f"Your current score is {score}/{n+1}")
            print("\n")
            print("-" * 10)
        else:
            print("That's wrong. !")
            print(f"The correct answer was {quiz_data[n]['answer']}")
            print(f"Your current score is {score}/{n+1}")
            print("\n")
            print("-" * 10)

    if current_question_num == total_questions:
        print(f"Finished! Final score: {score}/{total_questions}")
        break
```
