```python
print("What do you choose? Type 0 for Rock, 1 for Paper or 2 for Scissors")
a = int(input("What do you choose? Type 0 for Rock, 1 for Paper or 2 for Scissors"))

if a == 0:
    print("rock")
elif a == 1:
    print("paper")
elif a == 2:
    print("Scissors")
else:
    print("You pick wrong thing")

import random
b = random.randint(0,2)

print("Computer chose:")
if b == 0:
    print("rock")
elif b == 1:
    print("paper")
else:
    print("Scissors")

if a > b:
    print("You Win !!!")
else:
    print("You loose!!")
```
