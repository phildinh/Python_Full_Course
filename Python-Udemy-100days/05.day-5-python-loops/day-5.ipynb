```python
# Lowercase letters
letters = list("abcdefghijklmnopqrstuvwxyz")

# Numbers
numbers = list("0123456789")

# Symbols
symbols = list("!@#$%^&*()-_=+[]{}|;:,.<>?/`~")
```

```python
print("Welcome to the VScodePassword Generator!")
nr_letters = int(input("How many letters would you like in your password?\n"))
nr_symbols = int(input(f"How many symbols would you like?\n"))
nr_numbers = int(input(f"How many numbers would you like?\n"))

# we use random.choice() to get the letter
passwords = ""

for char in range(0, nr_letters):
    passwords += random.choice(letters)

for char in range(0, nr_numbers):
    passwords += random.choice(numbers)

for char in range(0, nr_symbols):
    passwords += random.choice(symbols)

print(passwords)
```

```python
print("Welcome to the VScodePassword Generator!")
nr_letters = int(input("How many letters would you like in your password?\n"))
nr_symbols = int(input(f"How many symbols would you like?\n"))
nr_numbers = int(input(f"How many numbers would you like?\n"))

# we use random.choice() to get the letter
passwords_list = []

for char in range(0, nr_letters):
    passwords_list.append(random.choice(letters))

for char in range(0, nr_numbers):
    passwords_list.append(random.choice(numbers))

for char in range(0, nr_symbols):
    passwords_list.append(random.choice(symbols))

print(passwords_list)
random.shuffle(passwords_list)
print(passwords_list)

passwords = ""
for char in passwords_list:
    passwords += char

print(f"This is your passwords {passwords}")
```

```python
print("Welcome to the VScodePassword Generator!")
nr_letters = int(input("How many letters would you like in your password?\n"))
nr_symbols = int(input(f"How many symbols would you like?\n"))
nr_numbers = int(input(f"How many numbers would you like?\n"))

# we use random.choice() to get the letter
passwords = ""

for char in range(0, nr_letters):
    passwords += random.choice(letters)

for char in range(0, nr_numbers):
    passwords += random.choice(numbers)

for char in range(0, nr_symbols):
    passwords += random.choice(symbols)

current_pass = list(passwords)
random.shuffle(current_pass)
print(current_pass)

new_passwords = ""
for char in current_pass:
    new_passwords += char

print(f"This is your passwords: {new_passwords}")
```
