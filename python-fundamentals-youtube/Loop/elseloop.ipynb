```python
# Check files if not a csv
files = [
    'data1.csv',
    'report.pdf',
    'report2.csv'
]

for file in files:
    if not file.endswith('.csv'):
        print(f'{file} is not a CSV')
        break
else:
    print('all files are good')
```

```python
# Python Challenge
file_list = [
    'report.csv',
    'data.xlsx',
    'summary.docsx',
    'report.csv',
    'data.csv'
]

if len(file_list) != len(set(file_list)):
    print("Duplicate found")
else:
    print("All files are unique")

seen = set()
duplicate_found = False

for file in file_list:
    if file in seen:
        duplicate_found = True
        break
    seen.add(file)

if duplicate_found:
    print("Duplicate found")
else:
    print("All files are unique")
```

```python
file_list = [
    'report.csv',
    'data.xlsx',
    'summary.docsx',
    'report.csv',
    'data.csv'
]

duplicates = set()

for file in file_list:
    if file_list.count(file) > 1:
        duplicates.add(file)

if duplicates:
    print("Duplicate found:")
    for d in duplicates:
        print(d)
else:
    print("All files are unique")
```
