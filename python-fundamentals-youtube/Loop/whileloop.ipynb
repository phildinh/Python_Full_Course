```python
# while true
attempts = 0
while attempts < 3:
    answer = input("Do you agree (yes/no): ")
    if answer == 'yes':
        print("Glad we are of the same page")
        break
    attempts += 1
else:
    print("3 strikes. you are out!!")
```
