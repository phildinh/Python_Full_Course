```python
# Logical Operations
print(3 > 1 and 5 < 10)  # True
print(3 > 1 or 5 > 10)   # True
print(3 < 1 or 5 > 10)   # False
print(not (3 > 1))       # False
print(not (3 < 1))       # True
```

```python
# use () for logical operators
# Allow access only if the user is logged in
# or they are a guest user
# but they must not be banned

is_logged_in = True
is_guest = False
is_banned = False

print((is_logged_in or is_guest) and not is_banned)  # True
```
