```python
# INdexing and Slicing
course = "Python Full Course"

# Extract first character
print(course[0])  # P
print(course[-len(course)])  # P

# Extract last character
print(course[-1])  # e
print(course[len(course)-1])  # e
```

```python
# Indexes & Slicing
date = "2024-06-15"

# Extract year
print(date[0:4])  # 2024
print(date[:4])  # 2024

# Extract month
print(date[5:7])  # 06

# Extract day
print(date[8:10])  # 15
print(date[-2:])  # 15
```

```python
# White space removal
text = "   Hello, World!   "
print(text.strip())  # "Hello, World!"
print(text.lstrip())  # "Hello, World!   "
print(text.rstrip())  # "   Hello, World!"
print(len(text))  # 19
print(len(text.strip()))  # 13
```

```python
# find example
phone1 = "+49 (176) 123-4567"
phone2 = "+1 (555) 987-6543"
phone3 = "+91 (987) 654-3210"

# lookup the position of the first space
print(phone1[4:])
print(phone2[3:])

# lookup the position of the first space
print(phone1.find(" "))
print(phone1[phone1.find(" ")+1 :])
print(phone2[phone2.find(" ")+1 :])
print(phone3[phone3.find(" ")+1 :])

# lookuo the position of the second space
first_space_index = phone1.find(" ")
print(first_space_index)
second_space_index = phone1.find(" ", first_space_index + 1)
print(second_space_index)
print(phone1[second_space_index + 1 :])  # 123-4567
print(phone2[second_space_index + 1 :])  # 123-4567
print(phone3[second_space_index + 1 :])  # 123-4567
```
