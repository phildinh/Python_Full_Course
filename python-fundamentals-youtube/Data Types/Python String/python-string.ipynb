```python
# Translation example
price = "123.45USD"
print(price.replace("USD", ""))  # 123.45VND

phone = "186-321-3122"
print(phone.replace("-", ""))  # 1863213122

price = "$1,234.56"
print(price.replace("$", "").replace(",", ""))  # 1234.56
```
