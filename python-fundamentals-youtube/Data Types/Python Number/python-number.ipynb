```python
# Numeric types example
int_var = 42
float_var = 3.14
complex_var = 1 + 2j

print(type(int_var))
print(type(float_var))
print(type(complex_var))
```

```python
# numeric types change
x = "3.14"
print(type(x))
print(float(x))
print(int(float(x)))
print(type(int(float(x))))
```

```python
# operators with numbers
a = 10
b = 3

print(a + b)  # Addition
print(a - b)  # Subtraction
print(a * b)  # Multiplication
print(a / b)  # Division
print(a // b) # Floor Division
print(a % b)  # Modulus
print(a ** b) # Exponentiation
```

```python
# shortcut operators
x = 5
x += 3  # x = x + 3

print(x)

x *= 2  # x = x * 2
print(x)
```

```python
# Measure Distance (Number rounding Example)
print(abs(2 - 5))  # Absolute value

# Rounding examples
print(round(3.14159, 2))  # Round to 2 decimal places
print(round(3.14159))     # Round to nearest integer
print(round(34.5123123123))
```

```python
import math
# Rounding using math module
price = 7.89

print(math.floor(price))  # Floor value
print(math.ceil(price))   # Ceiling value
print(round(price))
print(int(price))  # Truncate decimal part
```

```python
# Number functions random example
import random

print(random.random())
print(random.randint(1, 10))
```

```python
# Number functions validation example
x = 7.0
print(x.is_integer())

y = 7.1
print(y.is_integer())

x = 70.4
print(isinstance(x, int))
print(isinstance(x, float))
```

```python
# Python challenges with numbers
# Generate a random integer between 1 and 100 and check if it's even or odd
import random
random_number = random.randint(1, 100)
print(isinstance(random_number, int))
if random_number % 2 == 0:
    print(f"{random_number} is even.")
else:
    print(f"{random_number} is odd.")
```
