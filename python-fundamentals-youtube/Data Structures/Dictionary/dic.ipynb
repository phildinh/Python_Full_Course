```python
my_dict = {
    'a':10,
    'b':20,
    'c':30,
    'a':40
}

print(my_dict) #Ordered
# Keys are unique
# Values allow Duplicates
print(my_dict['b']) # Not Indexed(Keyed)
my_dict['c'] = 80
print(my_dict)
```

```python
user = {"id": 1, "age": 30, "city": "berlin"}

#Access
print(user.get("name","Unknow"))

#Checks
print("age" in user)
print("name" not in user)

#View Objects
print(user.keys())
print(user.values())
print(user.items())
```

```python
user = {"id": 1, "age": 30, "city": "berlin"}

# Looping
for u in user:
    print(u,user[u])

for key,value in user.items():
        print(key,value)
```

```python
user = {"id": 1, "age": 30, "city": "berlin"}

# Add, Remove, Update
user["name"] = "John"
user["age"] = 35
user.update({"age":40, "City":"Paris"})
print(user)

age = user.pop("age")
print(user)
print("Removed Item:", age)
```

```python
# Creation
user = {
    "id": None,
    "name": None,
    "age": None,
    "city": None
}

user = dict.fromkeys(["id","name","age","City"],0)
print(user)
```

```python
# Challenge: Keep only string values & Convert them to UPPERCASE
user = {"id": 1, "Name": "John", "age":30, "city": "Berlin"}

user_str = {
    #Expression
    k: v.upper()
    for k,v in user.items() #Loop
    if isinstance(v,str) #Filter
}

print(user_str)
```
