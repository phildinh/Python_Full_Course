```python
# Asterisk
person = ['Phil',29,'Data Analyst','city','Vietnam']

name, *details, country = person
print(name)
print(details)
print(country)
```

```python
# How to explore & Analyze
numbers = [1,5,5,2,3,4]

print("Max:",max(numbers))
print("Min:",min(numbers))
print("Sum:",sum(numbers))
print("Length:",len(numbers))

print("All:", all(numbers))
print("All:", all([1,0,'a']))

print("Any:", any(numbers))
print("Any:", any([1,0,'b']))

print("Count:", numbers.count(5))
print("Count:", numbers.count(6))

print("Index:", numbers.index(5))
print("Index:", numbers.index(5))

print(8 in numbers)
print(8 not in numbers)

list1 = [1,2,3]
list2 = [1,2,3]
print(list1 == list2)
```
