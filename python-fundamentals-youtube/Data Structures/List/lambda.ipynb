```python
multiple = lambda x: x * 2
print(multiple(2))
```

```python
add = lambda x,y: x + y
print(add(1,2))
```

```python
check = lambda i:i in "python"
print(check('n'))
```

```python
prices = ['$12.50','$9.99','$100.00']

p = '$12.50'
print(float(float(p.replace('$',""))))
```

```python
# lambda + map
prices = ['$12.50','$9.99','$100.00']
print(list(map(lambda p: float(p.replace('$','')),prices)))
```

```python
# lambda + filer
prices = [120, 30, 300, 80]

print(list(filter(lambda p: p>= 100,prices)))
```

```python
students = [
    ['Maria', 85],
    ['Kumar', 90],
    ['Max', 60]
]

print(list(filter(lambda row: row[1] > 70, students)))

```

```python
students = [
    ['Maria', 85],
    ['Kumar', 90],
    ['Max', 60]
]

print(list(filter(lambda row: row[0].startswith('M'),students)))
```

```python
domains = [
    'www.google.com',
    'openai.com',
    'localhost',
    'WWWW.DATAWITHBARAA.COM'
]

cleaned = [
    d.lower().replace('www','')
    for d in domains
    if '.' in d
]

print(cleaned)
```
