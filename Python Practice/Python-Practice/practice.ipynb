```python
a = [8,7,6,3,2]

odd = []
even = []

for i in a:
    print(i)


for x in range(0, len(a)):
    if x % 2 == 0:
        print(f"even {a[x]}")
        even.append(a[x])
    else:
        print(f"{a[x]} is odd")
        odd.append(a[x])

print("even number is: ", even)
print("odd number is: ", odd)
print(max(a))
print(min(a))
print(sorted(a, reverse = True)[2])

b = [3,8,10,15,27,30]

for z in range(0,len(b)):
    if b[z] == 27:
        print(f"we found the target number: ({z}) in a list")
```

```python
a = [8,7,6,3,2]

max = a[0]
min = a[0]

for x in a:
    if x > max:
        max = x
    if x < min:
        min = x

print(max)
print(min)
```
