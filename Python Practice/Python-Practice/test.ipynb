```python
def grade(score: int) ->str:
    if score >= 90 and score <= 100:
        return "A"
    elif score >= 80 and score < 90:
        return "B"
    elif score >= 70 and score < 80:
        return "C"
    elif score < 70 and score >= 0:
        return "D"
    else:
        return "Invalid"
```

```python
tests = [-1, 0, 69, 70, 85, 100, 101]
print([grade(s) for s in tests])
```

```python
employees = [
    {"id": 1, "name": "Ada", "dept": "Eng", "salary": 120, "active": True},
    {"id": 2, "name": "Bob", "dept": "Sales", "salary": 90, "active": True},
    {"id": 3, "name": "Cy", "dept": "Eng", "salary": 130, "active": False},
    {"id": 4, "name": "Dee", "dept": "HR", "salary": 80, "active": True},
    {"id": 5, "name": "Eli", "dept": "Sales", "salary": 110, "active": True},
]

departments = [
    {"dept": "Eng", "manager": "Grace"},
    {"dept": "Sales", "manager": "Lin"},
    {"dept": "HR", "manager": "Sam"},
]
```

```python
# A1: annual salary for Ada (single number)
ada_salary = next(e["salary"] for e in employees if e["name"] == "Ada")
annual_salary = ada_salary * 12
annual_salary
```

```python
# A2: summary string
summary = f"{employees[0]['name']} in {employees[0]['dept']} earns {employees[0]['salary']}"
print(summary)
```

```python
#B3: names with salary >= 100
names_100_plus = [r["name"] for r in employees if r["salary"] >= 100]
print(names_100_plus)
```

```python
# B4: active AND Sales
active_sales = [e["name"] for e in employees if e["active"] and e["dept"] == "Sales"]
print(active_sales)
```

```python
# B5: count inactive
inactive_count = sum(1 for e in employees if e["active"])
print(inactive_count)
```

```python
# C6: list of tuples (name, dept)
name_dept_pairs = [(e["name"],e["dept"]) for e in employees]
print(name_dept_pairs)
```

```python
# C7: dict salary_by_id
salary_by_id = {e["id"]:e["salary"] for e in employees}
print(salary_by_id)
```

```python
# D9: count_by_dept
count_by_dept = {}
for e in employees:
    dept = e["dept"]
    count_by_dept[dept] = count_by_dept.get(dept,0) + 1 
print(count_by_dept)
```

```python
# D9: total_summary_by_dept
total_summary_by_dept = {}
for e in employees:
    dept = e["dept"]
    total_summary_by_dept[dept] = total_summary_by_dept.get(dept,0) + e["salary"]
print(total_summary_by_dept)
```

```python
# D10: dept with highest total salary
top_dept = max(total_summary_by_dept,key=total_summary_by_dept.get)
print(top_dept)
```

```python
# E11: names sorted by salary desc
names_sorted_salary_desc = [e["name"] for e in sorted(employees, key=lambda x: x["salary"], reverse = True)]
print(names_sorted_salary_desc)
```

```python
# E12: top 2 earners
top2 = names_sorted_salary_desc[:2]
print(top2)
```

```python
# F13: manager_by_dept
manager_by_dept = {d["dept"]:d["manager"] for d in departments}
print(manager_by_dept)
```

```python
# F14: join rows
joined = [
    {"name": e["name"], "dept": e["dept"], "manager" : manager_by_dept.get(e["dept"])} 
     for e in employees
]
print(joined)
```

```python
# F15: join rows for active only
joined_active = [
    {"name": e["name"], "dept": e["dept"], "manager": manager_by_dept.get(e["dept"])}
    for e in employees
    if e["active"]
]
print(joined_active)
```
