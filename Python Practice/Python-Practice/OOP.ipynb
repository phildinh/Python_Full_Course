```python
class Worker:
    def __init__(self, name):
        self.name = name
    
    def introduce(self):
        print(f"Hi I am {self.name}")

class Engineer(Worker):
    def __init__(self, name,team):
        super().__init__(name)
        self.team = team
    
    def introduce(self):
        print(f"Hi I am {self.name} and I work in the {self.team}")


p1 = Worker("Phil")
print(p1.name)
p1.introduce()

p2 = Engineer("Phil","Data Team")
p2.introduce()
        
```

```python
class DataSource:
    def __init__(self, source_name):
        self.source_name = source_name
    def log(self, message):
        print(f"{self.source_name} {message}")
    def extract(self):
        print("extract() not implemented for this source")

class APISource(DataSource):
    def __init__(self, source_name, api_url):
        super().__init__(source_name)
        self.api_url = api_url
    def extract(self):
        print(f" Extracting from API: {self.api_url}")
    def validate_url(self):
        if self.api_url.startswith("https"):
            return True
        return False

class DatabaseSource(DataSource):
    def __init__(self, source_name, db_name):
        super().__init__(source_name)
        self.db_name = db_name
    def extract(self):
        print(f"Extracting from database: {self.db_name}")

p1 = DataSource("FakeStoreAPI")
p1.log("starting extraction")
p1 = APISource("FakeStoreAPI", "https://fakestoreapi.com/products")
p1.extract()
p1 = DatabaseSource("FakeStoreAPI", "PostgresSQL")
p1.extract()

# DataSource base
p1 = DataSource("GenericSource")
p1.log("starting extraction")
p1.extract()

# APISource
p2 = APISource("FakeStoreAPI", "https://fakestoreapi.com/products")
p2.extract()
p2.log("starting extraction")
print(p2.validate_url())

# DatabaseSource
p3 = DatabaseSource("PostgresDB", "retail_etl_dev")
p3.extract()
p3.log("starting extraction")
```
